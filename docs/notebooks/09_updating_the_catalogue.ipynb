# 09 — Updating the catalogue

Each `xeo` release includes a local snapshot of the Awesome Earth Observation Instruments catalogue. `Catalogue.update()` compares that snapshot with the current catalogue on GitHub and replaces the local JSON file only when the contents differ.

## Before updating

Updating requires an internet connection and write access to the installed `xeo/data` directory. It changes the catalogue used by the current Python environment, but it does not change the installed `xeo` package version.

```python
import xeo

print(f"xeo version: {xeo.__version__}")
print(f"Catalogue version before update: {xeo.catalogue.version}")
print(f"Number of instruments: {len(xeo.instruments)}")
```

## Download the current catalogue

The method reports whether the local catalogue was already current or was replaced by a newer upstream version.

```python
instruments = xeo.instruments
xeo.catalogue.update()
```

## Continue in the current session

A successful update refreshes the catalogue and shared instrument collection immediately. There is no need to restart Python or import `xeo` again.

```python
print(f"Catalogue version after update: {xeo.catalogue.version}")
print(f"Number of instruments: {len(xeo.instruments)}")
print("Shared collection preserved:", instruments is xeo.instruments)
```
